```python
# Chargement des packages
import numpy as np
import pandas as pd
```

```python
# Chargement des données
synthese = pd.read_csv("data/agribalyse_synthese.csv")

detail_etapes = pd.read_csv("data/agribalyse_detail_etapes.csv")
```

Feuillet "Synthese"

```python
# Dimensions du feuillet
print("Nombre de lignes:", synthese.shape[0])
print("Nombre de colonnes:", synthese.shape[1])

# Premières et dernières lignes
display(synthese.head(10))
display(synthese.tail(10))
```

```python
# Nettoyage des noms de colonnes
synthese.columns = (
    synthese.columns
    .str.replace("\xa0", " ", regex=False)
    .str.strip()
)

# Noms avant renommage
print("Avant :", synthese.columns.tolist())

# Renommage
renommage = {
    "Code AGB": "code_agb",
    "Code CIQUAL": "code_ciqual",
    "Groupe d'aliment": "groupe_aliment",
    "Sous-groupe d'aliment": "sous_groupe_aliment",
    "Nom du Produit en Français": "nom_produit",
    "LCI Name": "nom_lci",
    "code saison": "code_saison",
    "code avion": "code_avion",
    "Livraison": "livraison",
    "Approche emballage": "approche_emballage",
    "Préparation": "preparation",
    "DQR": "dqr",
    "Score unique EF": "score_unique_ef",
    "Changement climatique": "changement_climatique",
    "Appauvrissement de la couche d'ozone": "appauvrissement_ozone",
    "Rayonnements ionisants": "rayonnements_ionisants",
    "Formation photochimique d'ozone": "formation_photochimique_ozone",
    "Particules fines": "particules_fines",
    "Effets toxicologiques sur la santé humaine : substances non-cancérogènes":
        "toxicite_humaine_non_cancerogenes",
    "Effets toxicologiques sur la santé humaine : substances cancérogènes":
        "toxicite_humaine_cancerogenes",
    "Acidification terrestre et eaux douces":
        "acidification_terrestre_eaux_douces",
    "Eutrophisation eaux douces":
        "eutrophisation_eaux_douces",
    "Eutrophisation marine":
        "eutrophisation_marine",
    "Eutrophisation terrestre":
        "eutrophisation_terrestre",
    "Écotoxicité pour écosystèmes aquatiques d'eau douce":
        "ecotoxicite_eaux_douces",
    "Utilisation du sol":
        "utilisation_du_sol",
    "Épuisement des ressources eau":
        "epuisement_ressources_eau",
    "Épuisement des ressources énergétiques":
        "epuisement_ressources_energetiques",
    "Épuisement des ressources minéraux":
        "epuisement_ressources_mineraux",
    "Changement climatique - émissions biogéniques":
        "changement_climatique_biogeniques",
    "Changement climatique - émissions fossiles":
        "changement_climatique_fossiles",
    "Changement climatique - émissions liées au changement d'affectation des sols":
        "changement_climatique_changement_affectation_sols"
}

synthese = synthese.rename(columns=renommage)

# Noms après renommage
print("Après :", synthese.columns.tolist())

# Dimensions
print("Nombre de lignes :", synthese.shape[0])
print("Nombre de colonnes :", synthese.shape[1])
```

Feuillet "Detail etape"

```python
# Dimensions du feuillet
print("Nombre de lignes:", detail_etapes.shape[0])
print("Nombre de colonnes:", detail_etapes.shape[1])

# Premières et dernières lignes
display(detail_etapes.head(10))
display(detail_etapes.tail(10))
```

```python
# Extraction des noms des colonnes

colonnes_detail = detail_etapes.columns.tolist()

print("Les colonnes de la base détail sont :")
print(colonnes_detail)
```

```python
# Sélection des variables utiles

colonnes_a_conserver = [
    "Code AGB",
    "Changement climatique - Agriculture",
    "Changement climatique - Transformation",
    "Changement climatique - Emballage",
    "Changement climatique - Transport",
    "Changement climatique - Supermarché et distribution",
    "Changement climatique - Consommation",
]

detail_etapes = detail_etapes[colonnes_a_conserver].copy()

print("Nombre de colonnes après sélection :", detail_etapes.shape[1])

# Renommage des colonnes

renommage_detail = {
    "Code AGB": "code_agb",
    "Changement climatique - Agriculture": "changement_climatique_agriculture",
    "Changement climatique - Transformation": "changement_climatique_transformation",
    "Changement climatique - Emballage": "changement_climatique_emballage",
    "Changement climatique - Transport": "changement_climatique_transport",
    "Changement climatique - Supermarché et distribution": "changement_climatique_distribution",
    "Changement climatique - Consommation": "changement_climatique_consommation",
}

detail_etapes = detail_etapes.rename(columns=renommage_detail)

print("Colonnes après renommage :")
print(detail_etapes.columns.tolist())
```

Merge des deux feuillets

```python
# Fusion des tables Agribalyse

# Vérification des codes AGB
print("Nombre de codes AGB uniques dans synthese :", synthese["code_agb"].nunique())
print("Nombre de codes AGB répétés dans synthese :", synthese["code_agb"].duplicated().sum())

print("Nombre de codes AGB uniques dans detail_etapes :", detail_etapes["code_agb"].nunique())
print("Nombre de codes AGB répétés dans detail_etapes :", detail_etapes["code_agb"].duplicated().sum())

# Fusion des deux tables
agribalyse = synthese.merge(
    detail_etapes,
    on="code_agb",
    how="left",
    suffixes=("", "_detail")
)

# Vérification de la table obtenue
print("\nDimensions de la table Agribalyse :", agribalyse.shape)
print("\nPremières lignes :")
display(agribalyse.head())

# Enregistrement de la table Agribalyse
agribalyse.to_csv("data/agribalyse.csv", index=False)

print("\nTable Agribalyse enregistrée dans data/agribalyse.csv")
```
